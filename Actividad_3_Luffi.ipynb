```python
import pandas as pd
import matplotlib.pyplot as plt

ruta_csv = "/content/drive/MyDrive/Colab Notebooks/archivos para practica/Automobile.csv"
df = pd.read_csv(ruta_csv)
print("Vista previa del dataset:")
display(df.head())
print("Columnas disponibles:", df.columns.tolist())
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["weight"], df["mpg"], alpha=0.6, c="hotpink", edgecolors="navy")
plt.title("Relación entre peso del auto y consumo (mpg)", fontsize=14, color="navy")
plt.xlabel("Peso (lbs)", fontsize=12, color="navy")
plt.ylabel("Millas por galón (mpg)", fontsize=12, color="navy")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby("model_year")["mpg"].mean().plot(kind="line", marker="o", color="hotpink")
plt.title("Evolución del consumo (mpg) promedio por año", fontsize=14, color="navy")
plt.xlabel("Año del modelo", fontsize=12, color="navy")
plt.ylabel("Millas por galón (promedio)", fontsize=12, color="navy")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(8,6))
df.groupby("origin")["mpg"].mean().plot(kind="bar",
                                        color=["deeppink","royalblue","turquoise"])

plt.title("Promedio de consumo de combustible por región", fontsize=14, color="navy")
plt.xlabel("Región", fontsize=12, color="navy")
plt.ylabel("Millas por galón (mpg)", fontsize=12, color="navy")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(10,6))
for cyl, color in zip(sorted(df["cylinders"].unique()), ["deeppink","royalblue","turquoise","violet","lightblue"]):
    subset = df[df["cylinders"] == cyl]
    plt.hist(
        subset["mpg"],
        bins=15,
        alpha=0.5,
        label=f"{cyl} cilindros",
        color=color,
        edgecolor="navy"   # 🔑 borde azul marino
    )

plt.title("Distribución de mpg segmentada por cilindros", fontsize=14, color="navy")
plt.xlabel("Millas por galón (mpg)", fontsize=12, color="navy")
plt.ylabel("Frecuencia", fontsize=12, color="navy")
plt.legend(title="Cilindros")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
plt.figure(figsize=(6,5))
df.boxplot(column="mpg", by="cylinders", grid=False,
           boxprops=dict(color="deeppink"), whiskerprops=dict(color="royalblue"),
           capprops=dict(color="royalblue"), medianprops=dict(color="navy"))
plt.title("Consumo (mpg) según cantidad de cilindros", fontsize=14, color="deeppink")
plt.suptitle("")  # para quitar el subtítulo automático
plt.xlabel("Cilindros", fontsize=12, color="navy")
plt.ylabel("Millas por galón (mpg)", fontsize=12, color="navy")
plt.show()
```

```python
plt.figure(figsize=(6,5))
df.boxplot(column="mpg", by="cylinders", grid=False,
           showfliers=False,   # 🔑 elimina los circulitos
           boxprops=dict(color="deeppink"),
           whiskerprops=dict(color="royalblue"),
           capprops=dict(color="royalblue"),
           medianprops=dict(color="navy"))
plt.title("Consumo (mpg) según cantidad de cilindros", fontsize=14, color="deeppink")
plt.suptitle("")
plt.xlabel("Cilindros", fontsize=12, color="navy")
plt.ylabel("Millas por galón (mpg)", fontsize=12, color="navy")
plt.show()
```

```python
plt.figure(figsize=(12,6))
df.groupby(["model_year", "origin"]).size().unstack().plot(
    kind="bar", stacked=True, figsize=(12,6),
    color=["hotpink","royalblue","lightpink"], edgecolor="navy"
)
plt.title("Distribución de autos por año y origen", fontsize=14, color="navy")
plt.xlabel("Año del modelo", fontsize=12, color="navy")
plt.ylabel("Cantidad de autos", fontsize=12, color="navy")
plt.legend(title="Origen")
plt.show()
```
