# Ejercicio 1 - Importación y manipulación de datos con Pandas

En este ejercicio se importarán datos provenientes de archivos CSV, JSON y Excel.
Luego se realizarán operaciones básicas de exploración y se integrarán los datos en una única estructura.

```python
carpeta = "/content/drive/MyDrive/Colab Notebooks/archivos para practica/"
```

## 1. Carga de datos de ventas desde CSV

```python
ventas = pd.read_csv(carpeta + "ventas.csv")

print("Vista previa de los datos de ventas:")
display(ventas.head())
```

## 2. Carga de datos de clientes desde JSON

```python
clientes = pd.read_json(carpeta + "clientes.json")

print("Vista previa de los clientes:")
display(clientes.head())

print("Cantidad total de clientes:", len(clientes))
```

## 3. Carga de datos de inventario desde Excel

```python
inventario = pd.read_excel(carpeta + "inventario.xlsx")

print("Vista previa del inventario:")
display(inventario.head())
```

```python
promedio_stock = inventario["stock_disponible"].mean()

print("Promedio de stock disponible:", promedio_stock)
```

## 4. Integración de los conjuntos de datos

Los datos de ventas, clientes e inventario se combinan utilizando los identificadores comunes presentes en los archivos.

```python
ventas_clientes = pd.merge(
    ventas,
    clientes,
    left_on="id_cliente",
    right_on="id",
    how="left"
)

datos_completos = pd.merge(
    ventas_clientes,
    inventario,
    on="producto",
    how="left"
)

print("Datos combinados:")
display(datos_completos.head())
```

# Ejercicio 2 - Exploración y visualización de datos con Matplotlib

En este ejercicio se analizará el conjunto de datos Automobile.csv mediante distintas técnicas de visualización para identificar distribuciones, relaciones y posibles valores atípicos.

```python
autos = pd.read_csv(carpeta + "Automobile.csv")

print("Vista previa del conjunto de datos:")
display(autos.head())
```

## Relación entre peso y consumo de combustible

Se utiliza un gráfico de dispersión para observar la relación entre el peso del automóvil y las millas recorridas por galón de combustible.

```python
plt.figure(figsize=(8, 5))

plt.scatter(autos["weight"], autos["mpg"])

plt.xlabel("Peso del automóvil")
plt.ylabel("Millas por galón (mpg)")
plt.title("Relación entre peso y consumo de combustible")

plt.show()
```

Se observa una tendencia descendente: a medida que aumenta el peso de los automóviles, el valor de mpg tiende a disminuir. Esto sugiere que los vehículos más pesados presentan un mayor consumo de combustible.

```python
plt.figure(figsize=(8, 5))

plt.hist(autos["mpg"], bins=15, edgecolor="black")

plt.xlabel("Millas por galón (mpg)")
plt.ylabel("Frecuencia")
plt.title("Distribución del consumo de combustible")

plt.show()
```

El histograma permite observar cómo se distribuyen los valores de consumo de combustible y cuáles son los rangos de mpg más frecuentes dentro del conjunto de datos.

```python
print(autos[["cylinders", "origin"]].dtypes)

print("\nValores de cylinders:")
print(autos["cylinders"].unique())

print("\nValores de origin:")
print(autos["origin"].unique())
```

```python
autos_boxplot = autos.copy()

autos_boxplot["origin_num"] = autos_boxplot["origin"].map({
    "usa": 1,
    "europe": 2,
    "japan": 3
})

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        autos_boxplot["cylinders"],
        autos_boxplot["origin_num"]
    ],
    tick_labels=["Cilindros", "Origen"]
)

plt.ylabel("Valores")
plt.title("Diagrama de caja de cilindros y origen")

plt.show()
```

Para realizar el boxplot, la variable origin, originalmente categórica, fue codificada numéricamente como 1 = Estados Unidos, 2 = Europa y 3 = Japón. Dado que se trata de una variable categórica, su interpretación mediante un diagrama de caja es limitada, pero se realiza para cumplir con la consigna propuesta.

```python
distribucion = pd.crosstab(
    autos["model_year"],
    autos["origin"]
)

display(distribucion)
```

```python
distribucion.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución de automóviles según año y origen")
plt.legend(
    title="Origen",
    labels=["Estados Unidos", "Europa", "Asia"]
)

plt.show()
```

El gráfico permite comparar la cantidad de automóviles de cada origen a través de los distintos años del modelo, facilitando la identificación de cambios en su distribución a lo largo del tiempo.
